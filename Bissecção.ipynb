Resumo: ***O Método da Bissecção serve para encontrar raízes de funções. A ideia do método é escolher um extremo inferior $x_l$ e um superior $x_u$, que forma o intervalo que deve conter a raiz da função. Isto é garantido pela troca de sinais na função nesses extremos. Iterativamente, calculamos o ponto médio (que é a estimativa da raiz), verificamos se o erro é menor que a tolerância, se for paramos, senão verificamos se está na metade inferior ou superior, O método sempre converge, mas é lento***

```python
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Bibliotecas
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
import numpy as np                                                                                #biblioteca para funções matemáticas
import matplotlib.pyplot as plt                                                                   #blbiotecas de gŕaficos
import pandas as pd                                                                               #biblioteca de tabelas
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Entrada De Dados
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
m = 68.1                                                                                          #massa em kg
g = 9.8                                                                                           #gravidade em m/s²
v = 40                                                                                            #velocidade em m/s
t = 10                                                                                            #Instante de tempo
                                                                                                  #Função a ser encontrada a raiz:
def f(c):
    return(g * m / c ) * (1 - np.exp(- ( c / m ) * t )) - v

xl = 12                                                                                           #chute do extremo inferior do intervalo (chuta direito imundicie)
xu = 16                                                                                           #chute do extremo superior do intervalo (chuta direito imundicie)
tol = 1                                                                                           #tolerancia em porcentagem (não coloca o % aqui que vai dar erro)
iter_max = 100                                                                                    #numero máximo de interações
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                      #Implementação do Método
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------

def bisseccao(f, xl, xu, tol, iter_max):
    iter = 0                                                                                      #incializando o contador de iterações
    xr_velho = np.nan                                                                             #valor fictificio para inicializar o xr_velho
    erros = []                                                                                    #inicializando o vetor dos erros
    tabela = []                                                                                   #inicializando o tabela dos erros
    while iter < iter_max:                                                                        #enquanto não atingir o numero maximo de iterações
        xr_novo = (xl + xu) / 2                                                                   #calculando o ponto médio
        erro = np.abs((xr_novo - xr_velho) / xr_novo) * 100                                       #erro relativo percentual
        erros.append(erro)                                                                        #adicionando erro no vetor dos erros
        tabela.append({"n° da iteração": iter +1, "xr_novo": xr_novo, "erro(%)": erro})
        print(f"Na Iteração {iter + 1 }, a raiz aproximada é {xr_novo} e o erro é {erro}")
        if erro < tol:                                                                            #verificando se satisfez a tolerância
            return xr_novo, erros, tabela                                                                 #retornando o valor da raiz e os erros
        if f(xl) * f(xr_novo) < 0:                                                                #verificando se está no subintervalo inferido
            xu = xr_novo                                                                          #atualiza xu para a proxima iteração
        else:
            xl = xr_novo                                                                          #atualiza xl para a poxima iteração
        xr_velho = xr_novo                                                                        #atualiza o xr_velho para a proxima iteração
        iter = iter + 1                                                                           #atualiza o contador de iterações
    print("Não foi possível encontrar a raiz no número de iterações")
    return xr_novo, erros, tabela                                                                         #retorna o valor da raiz e os erros

#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Saída De Dados
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                                                                  #Chamar a função bissecção para encontrar a raiz da função
raiz, erros, tabela = bisseccao (f, xl, xu, tol, iter_max)
print(f"A raiz encontrada é {raiz}")
print (f"f(raiz) = {f(raiz)} (deve ser aproximadamente igual a zero)")

#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Saída De Dados OPCIONAIS (1)
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------

from matplotlib.lines import lineStyles
import numpy as np
import matplotlib.pyplot as plt

#Definindo a função f(c)
def f(c):
    return (667.38/c)* (1-np.exp(-0.146843 * c)) -40

c0 = 4
print (f"f({c0})=",f(4))

#Gerando valores de c no intervalo razoavel
x = np.linspace (10, 20, 100)
y = f(x)

#Criando Gŕafico
plt.figure (figsize=(10,6))
plt.plot(x, y, label='f(c)')
plt.axhline(0, color = 'red', linestyle = '--', linewidth=1)
plt.xlabel ('Coeficiente de Arrasto C')
plt.ylabel ('f(c)')
plt.title('Gráfico da função f(c) para encontrar o coerficiente de arrasto')
plt.grid(True)
plt.legend()
plt.show()
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Saída De Dados OPCIONAIS (4)
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
#--------------------------------------------------------------------------------------------------------------------------------------------------------------
                                                       #Gráfico do erro
iteracoes = range (1, len(erros) + 1)
plt.plot (iteracoes, erros)
plt.xlabel ("Número de Iterações")
plt.ylabel("Erro Relativo(%)")
plt.title("Gráfico do Erro para a Bissecção")
plt.grid(True)
plt.show()
```
